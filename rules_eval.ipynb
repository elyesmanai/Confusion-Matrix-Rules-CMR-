```python
import pandas as pd
from tqdm import tqdm
from sklearn.metrics import f1_score
import numpy as np
import explainer
import joblib

import warnings
warnings.filterwarnings("ignore")
```

```python
# KDD99 dataset
exclusive_rules = pd.read_csv('../Rules/KDD99_exclusive_rules.csv')
X_test = pd.read_csv('../Datasets/KDD99/X_test.csv')
y_test = pd.read_csv('../Datasets/KDD99/y_test.csv')
X_test['label'] = y_test
model = joblib.load('../Models/KDD99.pkl')
explainer.apply_exclusive_rules(X_test, exclusive_rules)
explainer.apply_exclusive_rules(X_test, exclusive_rules, model)
```

Resumee : 

SANS modèle:
- Couverture: 69.76%
- F1 sur couvert: 1.00 (parfait!)/ car sur notre zone de couverture il y a 100% de precision
- Non couverts: 30.24% (dont 0.08% conflits, 30.16% sans règle)
- F1 total (classe maj): 0.6916

AVEC modèle:
- Même couverture: 69.76%
- F1 sur couvert: 1.00
- F1 total (XGBoost): 0.9983 (Goooooood!)

Les règles exclusives sont parfaites là où elles s'appliquent. XGBoost est excellent pour compléter (F1=0.998)

et y a peu de conflits (0.08%) donc on conclu que les règles sont tres cohérentes
